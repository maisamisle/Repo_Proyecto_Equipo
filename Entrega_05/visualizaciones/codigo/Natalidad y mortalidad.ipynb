```python
import altair as alt
import pandas as pd

COLOR_HUESO = '#F1EDE6'
COLOR_NEGRO = '#191919'
COLOR_CENIZA = '#AAA6A1'

df_nat = pd.read_csv('01-base-de-datos-natalidad-limpia.csv', sep=';', decimal=',')
df_nat = df_nat.dropna().copy()
df_nat.columns = ['Territorio', '2023', '2024']

df_nat_long = df_nat.melt(id_vars='Territorio', var_name='Año', value_name='Tasa')

grafico_natalidad = alt.Chart(df_nat_long).mark_bar().encode(
    x=alt.X('Territorio:N', title=None, axis=alt.Axis(labelAngle=0)),
    y=alt.Y('Tasa:Q', title='Nacimientos por cada 1.000 habitantes'),
    color=alt.Color('Año:N', title='Año', scale=alt.Scale(
        domain=['2023', '2024'],
        range=[COLOR_CENIZA, COLOR_HUESO]
    )),
    xOffset='Año:N',
    tooltip=['Territorio', 'Año', alt.Tooltip('Tasa:Q', format='.2f')]
).properties(
    title='Tasa de natalidad: Israel vs. Ribera Occidental y Gaza (2023-2024)',
    width=450,
    height=320,
    background=COLOR_NEGRO
).configure_axis(
    labelColor=COLOR_HUESO,
    titleColor=COLOR_HUESO,
    gridColor='#333333',
    domainColor=COLOR_CENIZA
).configure_title(
    color=COLOR_HUESO,
    fontSize=15
).configure_legend(
    labelColor=COLOR_HUESO,
    titleColor=COLOR_HUESO
).configure_view(
    strokeWidth=0
)

display(grafico_natalidad)
```

```python
import altair as alt
import pandas as pd

COLOR_ROJO_SANGRE = '#9E1724'
COLOR_HUESO = '#F1EDE6'
COLOR_NEGRO = '#191919'
COLOR_CENIZA = '#AAA6A1'

df_mort = pd.read_csv('02-base-de-datos-mortalidad-limpia.csv', sep=';')
df_mort.columns = ['Territorio', '2023', '2024']

df_mort = df_mort[df_mort['Territorio'].isin(['Israel', 'State of Palestine'])].copy()
df_mort['Territorio'] = df_mort['Territorio'].replace({'State of Palestine': 'Palestina'})

df_mort_long = df_mort.melt(id_vars='Territorio', var_name='Año', value_name='Mortalidad')

grafico_mortalidad = alt.Chart(df_mort_long).mark_bar().encode(
    x=alt.X('Territorio:N', title=None, axis=alt.Axis(labelAngle=0)),
    y=alt.Y('Mortalidad:Q', title='Tasa de mortalidad'),
    color=alt.Color('Año:N', title='Año', scale=alt.Scale(
        domain=['2023', '2024'],
        range=[COLOR_CENIZA, COLOR_ROJO_SANGRE]
    )),
    xOffset='Año:N',
    tooltip=['Territorio', 'Año', 'Mortalidad']
).properties(
    title='Mortalidad comparada: Israel vs. Palestina (2023-2024)',
    width=450,
    height=320,
    background=COLOR_NEGRO
).configure_axis(
    labelColor=COLOR_HUESO,
    titleColor=COLOR_HUESO,
    gridColor='#333333',
    domainColor=COLOR_CENIZA
).configure_title(
    color=COLOR_HUESO,
    fontSize=15
).configure_legend(
    labelColor=COLOR_HUESO,
    titleColor=COLOR_HUESO
).configure_view(
    strokeWidth=0
)

display(grafico_mortalidad)
```
