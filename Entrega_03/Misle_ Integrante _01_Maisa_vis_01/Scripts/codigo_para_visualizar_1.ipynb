```python
import altair as alt
import pandas as pd

# Cargar el dataset
df = pd.read_csv("/content/ataques_hospitales_database_limpia_2.0.csv", sep=";")

# Extraer el año de la columna de fechas
df["Date_dt"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")
df["Year"] = df["Date_dt"].dt.year

# Obtener años y zonas existentes en el dataset
governorates = sorted(df["Governorate"].unique().tolist())
years = sorted(df["Year"].unique().tolist())

# Crear los selectores para interactividad (seleccionar zona y año)
input_gov = alt.binding_select(
    options=governorates,
    name="Gobernación: ",
)
select_gov = alt.selection_point(
    fields=["Governorate"],
    bind=input_gov,
    value=governorates[0],
)

input_year = alt.binding_select(
    options=years,
    name="Año: ",
)
select_year = alt.selection_point(
    fields=["Year"],
    bind=input_year,
    value=years[0],
)

# Crear grafico de barras
chart = (
    alt.Chart(df)
    .mark_bar()
    .encode(
        x=alt.X("count():Q", title="Cantidad de Ataques"),
        y=alt.Y("Type of attack:N", sort="-x", title="Tipo de Ataque"),
        color=alt.Color("Type of attack:N", legend=None),
        tooltip=[
            alt.Tooltip("Type of attack:N", title="Tipo de ataque"),
            alt.Tooltip("count():Q", title="Total de eventos"),
        ],
    )
    .add_params(select_gov, select_year)
    .transform_filter(select_gov)
    .transform_filter(select_year)
    .properties(
        width=550,
        height=320,
        title="Tipos de Ataques por Gobernación y Año",
    )
)

# Mostrar gráfico
chart
```
