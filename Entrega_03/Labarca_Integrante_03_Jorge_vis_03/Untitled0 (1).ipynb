```python
import altair as alt
import pandas as pd

```

```python
datos = pd.read_csv("killed_in_gaza_limpiado_3.csv", sep=";", encoding="utf-8")
datos.head()
```

```python
datos["grupo"].value_counts()
```

```python
alt.data_transformers.disable_max_rows()

alt.Chart(datos).mark_bar().encode(
    x="act:O",
    y="count()",
    color="grupo",
    tooltip=["act", "grupo", "count()"]
).interactive()
```

```python
alt.Chart(datos).mark_bar().encode(
    x=alt.X("grupo:N", title="Grupo", sort="-y"),
    y=alt.Y("count()", title="Número de personas"),
    color=alt.Color("grupo:N", legend=None),
    tooltip=[
        alt.Tooltip("grupo:N", title="Grupo"),
        alt.Tooltip("count()", title="Personas")
    ]
).properties(
    title="Personas por grupo en la base de datos",
    width=400,
    height=300
)
```

```python
ancho = 20

datos["n"] = datos.groupby(["act", "grupo"]).cumcount()
datos["col"] = datos["n"] % ancho + datos["grupo"].map({"BEBE": 0, "MADRE POTENCIAL": ancho + 2})
datos["fila"] = datos["n"] // ancho
```

```python
alt.data_transformers.disable_max_rows()

grafico = alt.Chart(datos[["act", "grupo", "col", "fila", "total"]]).mark_circle(size=6, opacity=1).encode(
    x=alt.X("col:Q", axis=None),
    y=alt.Y("fila:Q", title="Personas",
            axis=alt.Axis(labelExpr=f"datum.value * {ancho}", grid=False)),
    color=alt.Color("grupo:N",
                    scale=alt.Scale(domain=["BEBE", "MADRE POTENCIAL"],
                                    range=["#3ddc84", "#ffffff"]),
                    legend=alt.Legend(title="Grupo")),
    tooltip=[alt.Tooltip("act:O", title="Actualización"),
             alt.Tooltip("grupo:N", title="Grupo"),
             alt.Tooltip("total:Q", title="Total de personas")]
).properties(
    width=110, height=300
).facet(
    alt.Facet("act:O", title="Actualización"), columns=10
).properties(
    title="Bebés y madres potenciales fallecidas en Gaza por actualización cronológica"
).configure(
    background="#111111"
).configure_view(
    stroke=None
).configure_axis(
    labelColor="white", titleColor="white"
).configure_title(
    color="white"
).configure_header(
    labelColor="white", titleColor="white",
    labelOrient="bottom", titleOrient="bottom"
).configure_legend(
    labelColor="white", titleColor="white"
)

grafico
```

```python
datos["total"] = datos.groupby(["act", "grupo"])["n"].transform("count")
```
