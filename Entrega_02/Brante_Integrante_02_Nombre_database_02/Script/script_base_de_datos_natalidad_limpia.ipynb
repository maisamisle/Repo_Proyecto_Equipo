```python
import pandas as pd

# 1. Cargar el CSV
df = pd.read_csv('base-de-datos-natalidad-limpia.csv', sep=None, engine='python')

# 2. Tomar la primera fila (donde están los años real) como cabecera de las columnas
nuevas_columnas = df.iloc[0].astype(str).str.replace('-', '').str.strip()
df.columns = nuevas_columnas

# 3. Eliminar la primera fila (ya que la promovimos a encabezado)
df = df.iloc[1:].reset_index(drop=True)

# 4. Asignar 'País' a la primera columna
df = df.rename(columns={df.columns[0]: 'País'})

# 5. Establecer la columna 'País' como el índice limpio
df_final = df.set_index('País')

# 6. Mostrar la tabla perfecta
df_final
```
