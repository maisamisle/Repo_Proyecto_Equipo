```python
import pandas as pd

# 1. Cargar el CSV directamente
df_mortalidad = pd.read_csv('base-de-datos-mortalidad-limpia.csv', sep=None, engine='python')

# 2. Asignar 'País' a la primera columna
df_mortalidad = df_mortalidad.rename(columns={df_mortalidad.columns[0]: 'País'})

# 3. Eliminar columnas innecesarias (como Unnamed) o filas vacías (NaN)
df_mortalidad = df_mortalidad.loc[:, ~df_mortalidad.columns.str.contains('^Unnamed')]
df_mortalidad = df_mortalidad.dropna(how='all')

# 4. Limpiar los nombres de las columnas para que queden solo los años sin decimales (ej. 1998 en vez de 1998.0)
df_mortalidad.columns = [str(col).replace('.0', '').replace('-', '').strip() for col in df_mortalidad.columns]

# 5. Establecer 'País' como índice
df_mortalidad = df_mortalidad.set_index('País')

# 6. Mostrar la tabla impecable
df_mortalidad

```
